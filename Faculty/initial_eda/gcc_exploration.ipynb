The main goal was to see if this dataset held additional data that Transparent California did not include. I was told to look for things like Tenure Status and Positions of Faculty

```python
import polars as pl
import duckdb

conn = duckdb.connect()

# Read combined parquet file
gcc_df = pl.read_parquet("../data/GCC.parquet")

transparent_df = pl.read_parquet("../data/faculty-salaries.parquet")

```

```python
# We really only care about faculty data
df = conn.sql(""" 
        SELECT * FROM gcc_df WHERE Position ILIKE '%professor%'
                            OR Position ILIKE '%lecturer%'
                            OR Position ILIKE '%faculty%'
                            OR Position ILIKE '%instructor%'
        """).pl()

df.sample(25)
```

```python
# From the random sampling, it seems a lot of DeptOrSubdivision is missing, lets see how much stuff is missing in general
for col in df.columns:
    print(f"Percent of data missing from {col}: {((df[col].null_count() / df.height) * 100):.2f}%")
```

I am going to see if maybe we can join Transparent California and this GCC dataset, using TotalWages and totalpaybenefits. I would assume that this is how TransparentCalifornia does it  

```python
combined_df = conn.sql("""
    WITH gcc_agg AS (
        SELECT EmployerName, Position, TotalWages, Year
        FROM gcc_df
    ),
    transparent_agg AS (
        SELECT department, job, totalpay, year
        FROM transparent_df
        WHERE year = 2024
          AND department IS NOT NULL
    )
    SELECT
        g.EmployerName,
        g.Position,
        g.TotalWages,
        g.Year,
        t.department,
        t.job,
        t.totalpay
    FROM transparent_agg t
    JOIN gcc_agg g
        ON g.totalWages = t.totalpay
""").pl()

print(f"Matched {combined_df.height} rows")
combined_df.head(25)
```
